```python
import sqlite3
import pandas as pd
from datetime import datetime
```

```python
connection = sqlite3.connect("blog.db")
cursor = connection.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS blog_posts(
id INTEGER PRIMARY KEY AUTOINCREMENT,
title TEXT,
content TEXT,
author TEXT,
date TEXT
)
""")

connection.commit()
```

```python
def add_blog(title, content, author):

    date = datetime.now().strftime("%d-%m-%Y %H:%M:%S")

    cursor.execute("""
    INSERT INTO blog_posts(title, content, author, date)
    VALUES (?, ?, ?, ?)
    """, (title, content, author, date))

    connection.commit()

    print("Blog Added Successfully")
```

```python
add_blog(
    "Python Basics",
    "Python is an easy programming language.",
    "Karuna"
)

add_blog(
    "Machine Learning",
    "Machine learning is a subset of Artificial Intelligence.",
    "Karuna"
)
```

```python
cursor.execute("SELECT * FROM blog_posts")

rows = cursor.fetchall()

df = pd.DataFrame(
    rows,
    columns=[
        "ID",
        "Title",
        "Content",
        "Author",
        "Date"
    ]
)

df
```

```python
keyword = "Python"

cursor.execute("""
SELECT * FROM blog_posts
WHERE title LIKE ?
OR content LIKE ?
""",('%'+keyword+'%','%'+keyword+'%'))

result = cursor.fetchall()

pd.DataFrame(
    result,
    columns=[
        "ID",
        "Title",
        "Content",
        "Author",
        "Date"
    ]
)
```

```python
cursor.execute("""
UPDATE blog_posts
SET content=?
WHERE id=1
""",("Python is one of the easiest programming languages.",))

connection.commit()

print("Updated Successfully")
```
